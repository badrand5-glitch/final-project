```python
# 1. Import required Python libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configure plot styling
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 11

print(" All required libraries imported successfully!")

# 2. Load and explore the dataset
file_path = 'Copy of AB_NYC_2019.xlsx'
df = pd.read_excel(file_path)

print(f"Dataset Dimensions: {df.shape[0]} rows and {df.shape[1]} columns.\n")
print("--- First 5 Rows ---")
display(df.head())

# 3. Understand columns, non-null counts, and data types
print("--- Dataset Information ---")
df.info()

print("\n--- Summary Statistics (Numerical Columns) ---")
display(df.describe())

print("\n--- Summary Statistics (Categorical Columns) ---")
display(df.describe(include=['object']))

# 4. Identify and handle missing values and duplicates
duplicates = df.duplicated().sum()
print(f" Duplicate rows count: {duplicates}")

if duplicates > 0:
    df = df.drop_duplicates()
    print(" Duplicates dropped successfully.")

print("\n Missing values per column:")
missing_vals = df.isnull().sum()
print(missing_vals[missing_vals > 0])

df['name'] = df['name'].fillna('Unknown')
df['host_name'] = df['host_name'].fillna('Unknown')
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
df['last_review'] = df['last_review'].fillna(pd.to_datetime('2000-01-01'))


print("\n Missing values handled! Remaining missing values:", df.isnull().sum().sum())

# 5. Clean and prepare data for analysis
df['last_review'] = pd.to_datetime(df['last_review'])
df['min_revenue_potential'] = df['price'] * df['minimum_nights']

print("\n--- Cleaned Dataset Sample ---")
display(df[['id', 'name', 'neighbourhood_group', 'room_type', 'price', 'minimum_nights', 'min_revenue_potential']].head())

#  Charts and Visualizations 

# Chart 1: Distribution of Listings
plt.figure(figsize=(10, 5))
ax = sns.countplot(data=df, x='neighbourhood_group', order=df['neighbourhood_group'].value_counts().index, palette='viridis')
plt.title("Number of Listings per Neighbourhood Group", fontsize=14, fontweight='bold')
plt.xlabel("Neighbourhood Group", fontsize=12)
plt.ylabel("Total Listings Count", fontsize=12)
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='baseline', fontsize=10, color='black', xytext=(0, 5), textcoords='offset points')
plt.tight_layout()
plt.show()

# Chart 2: Average Price by Neighbourhood & Room Type
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='neighbourhood_group', y='price', hue='room_type', ci=None, palette='magma')
plt.title("Average Listing Price by Neighbourhood Group & Room Type", fontsize=14, fontweight='bold')
plt.xlabel("Neighbourhood Group", fontsize=12)
plt.ylabel("Average Price ($)", fontsize=12)
plt.legend(title="Room Type")
plt.tight_layout()
plt.show()

# Chart 3: Distribution of Room Types
plt.figure(figsize=(7, 7))
room_counts = df['room_type'].value_counts()
plt.pie(room_counts, labels=room_counts.index, autopct='%1.1f%%', startangle=140, colors=['#66b3ff','#99ff99','#ffcc99'], explode=(0.05, 0.05, 0.1))
plt.title("Share of Room Types Across NYC Listings", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Chart 4: Correlation Heatmap
plt.figure(figsize=(10, 6))
numeric_cols = ['price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month', 'calculated_host_listings_count', 'availability_365']
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix Heatmap", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## Answers to Key Business Questions:

1. **Which Neighbourhood Group has the highest number of listings and highest average prices?**
   - **Manhattan** leads in total listings (**21,661 listings**) and has the highest average price (**$196.88 / night**).
   - **Brooklyn** comes second with **20,104 listings** and an average price of **$124.38 / night**.

2. **What is the most common room type offered in NYC?**
   - **Entire home/apt** represents the majority of listings (**52.0%**), followed closely by **Private room** (**45.7%**). **Shared rooms** represent a tiny fraction (**2.4%**).

3. **Is there a relationship between price and review volume?**
   - There is a weak negative correlation (**-0.05**) between price and total reviews. Lower-priced listings tend to receive higher review volumes due to higher booking frequency and occupancy.

4. **Which top 5 neighbourhoods have the highest concentration of listings?**

## Key Findings & Summary

2. **Pricing Disparities:** Manhattan is significantly more expensive than other boroughs across all room types (average **$211.79** for Entire homes/apts vs **$70.13** for Shared rooms overall).
3. **Availability & Hosting Patterns:** Hosts with multiple listings (`calculated_host_listings_count`) tend to maintain higher year-round availability (`availability_365`).
